## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python


# Define some custom errors to let validate_read except or report them

class WrongLengthError(ValueError):
  """Raised when a sequence length does not match reference length."""

class InvalidCharacterError(ValueError):
  """Raised when a sequence contains a character other than DNA bases: A, T, G, C."""


# Helper functions

# Sequences need to be uppercase so they can be compared to reference base later on, as python is case sensitive.
# Removing flanking whitespace so otherwise valid sequences won't be mislabled as invalid
def clean_format_sequence(sequence: str) -> str:
  """
  Formats sequence by converting to uppercase and stripping flanking whitespace.

  Args:
      sequence: A string of any length

  Returns:
      A cleaned sequence with text converted to uppercase and surrounding whitespace removed"""
  return sequence.upper().strip()


# Includes error message if you wanted to check a sequence outside of validate_read
def is_dna(sequence: str) -> str:
  """
  Checks if a 'cleaned' sequence is DNA containing A/T/G/C only (case sensitive).

  Args:
      sequence (str): A 'cleaned' sequence returned by clean_format_sequence.

  Returns:
      sequence (str) : Input sequence if all characters are DNA bases

  Raises:
      InvalidCharacterError: If any characters aren't DNA bases"""

  for base in sequence:
    if base not in "ATGC":
        raise InvalidCharacterError(f"{sequence!r} contains {base!r} which isn't DNA base A/T/G/C")

  return sequence


def validate_read(sequence: str, reference: str) -> dict[str, str, str, bool, str, list]:
    """
    Validate a single read by cleaning the input sequence and comparing it against the input reference
    to check if cleaned sequence is non-zero, matches reference length and is DNA.
    Length and invalid character errors are excepted and are appended to errors list. A valid read has 0 errors.

    Args:
       sequence: A string of any length
       reference: A non-empty string of only valid DNA bases.

    Returns:
       A dict reporting the cleaned sequence, if validity is True/False, and an errors list """

    # Makes a list we can keep errors in, converts input sequence into a clean format.
    errors = []
    clean_seq = clean_format_sequence(sequence)

    # Checks if clean sequence length is over 0, and if so checks if both lengths match.
    # Catches custom errors and appends error type to errors[]
    try:
      if len(clean_seq) == 0:
        errors.append("wrong_length")

      if len(clean_seq) >0 and len(clean_seq)!= len(reference):
          raise WrongLengthError(f"Sequence length:{len(sequence)} doesn't match reference length: {len(reference)}")
    except WrongLengthError:
      errors.append("wrong_length")

  # Checks if clean sequence contains only DNA bases
    try:
        is_dna(clean_seq)
    except InvalidCharacterError:
        errors.append("invalid_characters")

    # A clean sequence can only be valid=True if error list is empty
    valid = len(errors) == 0

    return {"sequence": clean_seq, "valid": valid, "errors": errors}


````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python

def find_mutations(sequence: str, reference: str) -> list[dict]:
  """
  For each position (first base = position 1), checks for mismatches between sequence and reference.
  If no mutations are present, returned list will be empty.

  Args:
       sequence (str): A valid DNA sequence of matching length to reference.
       reference (str): A valid DNA reference you want to check sequence against.

    Returns:
       found_mutations: A list containing a dict per mutation found at a position"""

  found_mutations = []

  # Makes a dict and lines up each sequence and reference base with their
  # position number, with the first base being position 1.
  for i, (sequence_base, reference_base) in enumerate(zip(sequence, reference), start=1):

    # If the bases at a position dont match, append the details to found_mutations
    if sequence_base != reference_base:
      found_mutations.append({"position": i, "reference_base": reference_base, "read_base": sequence_base})

  return found_mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python

def analyse_reads(reads: list[dict], reference: str) -> list[dict]:
    """
    Analyses each read against the reference to analyse each read for validity status, errors, and mutations.

   Args:
       reads: A list of dicts, with each dict in this format:
              {"read_id": str,
               "sample": str,
               "group": str,
               "sequence": str}
       reference: A valid DNA reference you want to check sequence against.

    Returns:
       analysed_reads: A list containing a dict of results for each read"""

    analysed_reads = []

    for read in reads:
      seq = read.get("sequence")

      # Ensures function uses only validated sequences for accuracy
      validated_seq = validate_read(seq, reference)
      if validated_seq["valid"]:
        status = "valid"
      else: status = "invalid"

      # If sequence is valid, report the mutations, otherwise mutations is an empty list
      if status == "valid":
        seq_mutation = find_mutations(validated_seq.get("sequence"), reference)
      else: seq_mutation = []

      # Create results dictionary that ascribes the relevant values to the keys
      results = {
              "read_id": read.get("read_id"),
              "sample": read.get("sample"),
              "group": read.get("group"),
              "sequence": validated_seq.get("sequence"),
              "status": status,
              "errors": validated_seq.get("errors"),
              "mutations": seq_mutation}

      #Adds results for each read to analysed_reads list
      analysed_reads.append(results)

    return analysed_reads


````

````python


def summarise_samples(analysed_reads=None) -> dict[dict]:
    """
    Uses the results returned by analyse_reads() and creates a nested dict for each unique sample that
    summarises all reads within that sample, that reports group, valid reads, invalid reads, mismatch reads,
    mismatch percentage, mismatch events, substitution counts, position counts and top positions.

    Args:
    analysed_reads: The list of dictionaries returned from analyse_reads()

    Returns:
    samples_summary: A nested dict for each unique sample within analysed_reads
    {}: A empty dict if no parameter is given
    """

    # An empty input parameter will return {}
    if analysed_reads is None:
      return {}

    samples_summary = {}

    for read in analysed_reads:
      sample_type = read["sample"]
      group_type = read["group"]

      # Intialise dictionary for each unique sample type
      if sample_type not in samples_summary:
              samples_summary[sample_type] = {
              "group": group_type,
              "valid_reads": 0,
              "invalid_reads": 0,
              "mismatch_reads": 0,
              "mismatch_percentage": None,
              "mismatch_events": 0,
              "substitution_counts": {},
              "position_counts": {},
              "top_positions": []}

      # Count every valid read
      if read["status"] == "valid":
        samples_summary[sample_type]["valid_reads"] += 1

        # If valid, check the following:

        # Number of mismatch reads (has atleast 1 mismatch) and calculate mismatch percentage
        mismatch_reads = read["mutations"]
        if len(mismatch_reads) != 0:
          samples_summary[sample_type]["mismatch_reads"] += 1
          samples_summary[sample_type]["mismatch_percentage"] = (samples_summary[sample_type]["mismatch_reads"] / samples_summary[sample_type]["valid_reads"]) * 100

        # Mismatch events (number of mismatches in a read)
          for mismatch in mismatch_reads:
            samples_summary[sample_type]["mismatch_events"] +=1

            # Substitution count
            substitution_type = f"{mismatch['reference_base']}>{mismatch['read_base']}"
            substitution_count = samples_summary[sample_type]["substitution_counts"]
            substitution_count[substitution_type] = substitution_count.get(substitution_type, 0) + 1

            # Position count
            position_num = mismatch["position"]
            position_count = samples_summary[sample_type]["position_counts"]
            position_count[position_num] = position_count.get(position_num, 0) + 1


      elif read["status"] == "invalid":
          samples_summary[sample_type]["invalid_reads"] += 1

    # Top 3 positions from highest frequency
    for sample_type, name in samples_summary.items():
      rev_sorted_position_counts = sorted(name["position_counts"].items(), key=lambda kv: kv[1], reverse=True)
      name["top_positions"] = [position_num for position_num, i in rev_sorted_position_counts[:3]]

     # Sort position_counts to display in ascending key(position) order
     # (N.B. I'm sure theres a more elegant method, but I tried so many ideas and this worked ¯\_(ツ)_/¯)
      sorted_position_counts = dict(sorted(name["position_counts"].items(), key=lambda kv: kv[0], reverse=False))
      name["position_counts"] = sorted_position_counts


    return samples_summary


test_summary = summarise_samples(test_analysed_reads)
#print(test_summary)

real_summary = summarise_samples(analysed_real_reads)
print(real_summary)
````

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python

```

Each list must contain **one percentage per biological sample**.

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````
